This program demonstrates a simple LangChain + OpenAI interaction.
It uses a pre-configured language model to answer a basic question
about the capital of France and prints the response in the notebook.

```python
from langchain_openai import OpenAI
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
llm = OpenAI(model="gpt-3.5-turbo-instruct")
result = llm.invoke("What is the capital of France?")
print(result)
```
